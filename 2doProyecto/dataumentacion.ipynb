```python
from PIL import Image, ImageEnhance
import os

carpeta_entrada = "DatasetFormaciones"
carpeta_salida = "DatasetAumentado"

os.makedirs(carpeta_salida, exist_ok=True)

for clase in range(1,16):

    origen = os.path.join(carpeta_entrada, str(clase))
    destino = os.path.join(carpeta_salida, str(clase))

    os.makedirs(destino, exist_ok=True)

    for archivo in os.listdir(origen):

        if archivo.lower().endswith(('.jpg','.jpeg','.png')):

            ruta = os.path.join(origen, archivo)

            img = Image.open(ruta).convert("RGB")

            nombre = os.path.splitext(archivo)[0]

            # Guardar original
            img.save(f"{destino}/{nombre}_orig.jpg")

            # Rotación +5
            img.rotate(5).save(
                f"{destino}/{nombre}_rot5.jpg"
            )

            # Rotación -5
            img.rotate(-5).save(
                f"{destino}/{nombre}_rotm5.jpg"
            )

            # Más brillo
            ImageEnhance.Brightness(img)\
                .enhance(1.3)\
                .save(
                    f"{destino}/{nombre}_bright.jpg"
                )

            # Menos brillo
            ImageEnhance.Brightness(img)\
                .enhance(0.7)\
                .save(
                    f"{destino}/{nombre}_dark.jpg"
                )
```

```python
from PIL import Image
import numpy as np
import pandas as pd
import os
import torch

# Carpeta principal donde están las 15 clases
carpeta_base = "DatasetAumentado"

datos = []
etiquetas = []

# ==========================
# Procesar las 15 clases
# ==========================
for clase in range(1, 16):

    carpeta_clase = os.path.join(carpeta_base, str(clase))

    if not os.path.exists(carpeta_clase):
        print(f"No existe la carpeta: {carpeta_clase}")
        continue

    print(f"Procesando clase {clase}...")

    for archivo in os.listdir(carpeta_clase):

        if archivo.lower().endswith(('.jpg', '.jpeg', '.png')):

            ruta = os.path.join(carpeta_clase, archivo)

            try:
                img = Image.open(ruta)

                # Convertir a RGB
                img = img.convert('RGB')

                # Redimensionar
                img = img.resize((100, 100))

                # Convertir a vector
                vector = np.array(img).flatten()

                datos.append(vector)

                # Etiqueta de la clase
                # Si quieres clases desde 0 a 14:
                etiquetas.append(clase - 1)

                # Si prefieres clases desde 1 a 15:
                # etiquetas.append(clase)

            except Exception as e:
                print(f"Error con {ruta}: {e}")

# ==========================
# Crear dataset
# ==========================
X = np.array(datos, dtype=np.float32)

X = X / 255.0

df = pd.DataFrame(X)

df["clase"] = etiquetas

df.to_csv("dataset_formaciones_15_clases.csv", index=False)

print("\nCSV generado correctamente")
print("Cantidad de imágenes:", len(df))
print("Número de características:", X.shape[1])
print("Número de clases:", len(set(etiquetas)))
```
